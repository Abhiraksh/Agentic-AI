````python
from litellm import completion

def generate_response(messages):

  response = completion( # Has 3 relevant arguments: model (Claude, ChatGPT, etc.), message (role and prompt) and max_tokens
      model = "openai/gpt-4o",
      messages = messages,
      max_tokens = 500
  )
  return response.choices[0].message.content

def parse_code(response):

  if "```python" in response:

    code = response.split("```python")[1]
    code = code.split("```")[0]
    return code.strip()

  else:

    return response
````
